```python
import pandas as pd
```

```python
df = pd.read_excel('data.xlsx')
df = df.drop(['Nationality','Notes','Reference'],axis=1)
df = df.rename(columns={'#':'Selections',"Selections[c]":'Years'})

df["Player"] = (df["Player"].str.replace(r"\[.*?\]", "", regex=True).str.replace(r"[*^]", "", regex=True).str.strip())
df.to_csv('all_stars_clean.csv',index=False)
```
